```python
#Aktivitas 6F
skor = [80, 65, 90, 75, 70]

total = 0                    # akumulator: mulai dari nol
jumlah_lulus = 0             # penghitung = akumulator yang menambah 1
for nilai in skor:
    total = total + nilai    # tambahkan setiap elemen
    if nilai >= 75:
        jumlah_lulus = jumlah_lulus + 1
    print("total sementara:", total)

rata_rata = total / len(skor)
print("Total     :", total)
print("Rata-rata :", rata_rata)
print("Skor >= 75:", jumlah_lulus, "dari", len(skor))
```

```python
#Aktivitas 6G
suhu = [28, 31, 30, 33, 29, 27, 32]

# Perubahan harian: mulai dari indeks 1 karena kita melihat i - 1
for i in range(1, len(suhu)):
    perubahan = suhu[i] - suhu[i - 1]
    print(f"hari {i + 1}: {suhu[i]} °C, perubahan {perubahan:+d}")

# Rata-rata bergerak 3 hari: jendela = 3 nilai yang berakhir di indeks i
rata_3 = []
for i in range(2, len(suhu)):
    jendela = suhu[i - 2 : i + 1]
    rata_3.append(round(sum(jendela) / 3, 1))

print("Rata-rata 3 hari:", rata_3)
print(len(suhu), "nilai ->", len(rata_3), "rata-rata")

#ubah blok menjadi rata-rata 5 hari
suhu = [28, 31, 30, 33, 29, 27, 32]

rata_5 = []
for i in range(4, len(suhu)):
    jendela = suhu[i - 4 : i + 1]
    rata_5.append(round(sum(jendela) / 5, 1))

print("Rata-rata 5 hari:", rata_5)
print(len(suhu), "nilai →", len(rata_5), "rata-rata")
```

```python
#Pemanasan
skor = [80, 65, 90, 75, 70, 85]

#1.skor pertama dan terakhir
print("Pertama:", skor[0], " Terakhir:", skor[-1])

#2.rata-rata
rata = sum(skor) / len(skor)
print("Rata-rata:", rata)

#3.berapa skor di atas rata-rata
jumlah_atas = 0
for nilai in skor:
    if nilai > rata:
        jumlah_atas = jumlah_atas + 1
print("Di atas rata-rata:", jumlah_atas)

#4.posisi skor terendah sebagai nomor mahasiswa (mulai dari 1)
terendah = min(skor)
print("Skor terendah milik mahasiswa nomor", skor.index(terendah) + 1)

#5.list baru, setiap skor ditambah 5
skor_baru = []
for nilai in skor:
    skor_baru.append(nilai + 5)
print("Skor + 5:", skor_baru)
```

```python
#Latihan Praktikum
energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

#1.Total dan rata-rata
total = sum(energi)
rata = total / len(energi)
print("Total :", total)
print("Rata-rata:", round(rata, 1))

#2.Tiga hari tertinggi dan tiga terendah (nomor hari + nilai)
pasangan = []
for i in range(len(energi)):
    pasangan.append((energi[i], i + 1))   # (nilai, nomor hari)
urut = sorted(pasangan)
print("3 terendah :", [(hari, nilai) for nilai, hari in urut[:3]])
print("3 tertinggi:", [(hari, nilai) for nilai, hari in urut[-3:]])

#3.Jumlah hari di bawah target (diinput)
target = float(input("Masukkan target: "))
di_bawah = 0
for nilai in energi:
    if nilai < target:
        di_bawah = di_bawah + 1
print("Hari di bawah target:", di_bawah)

#4.Hari yang rata-rata bergerak 7 harinya < 97% rata-rata bulanan
batas = 0.97 * rata
print("Hari dengan rata-rata 7 hari di bawah 97%:")
for i in range(6, len(energi)):
    jendela = energi[i - 6 : i + 1]
    rata_7 = sum(jendela) / 7
    if rata_7 < batas:
        print("  hari", i + 1, "->", round(rata_7, 1))

#5.Total per minggu (minggu 1-4, sisa hari ditambahkan)
for m in range(4):
    awal = m * 7
    akhir = awal + 7
    if m == 3:
        akhir = len(energi)       # minggu 4 ikut menyerap sisa 2 hari (29-30)
    print("Minggu", m + 1, ":", sum(energi[awal:akhir]))
```
