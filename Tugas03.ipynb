```python
# Nama : Cindy Agatha Purba
# NIM : 4232501026
# Tanggal : 23 September 2026
# Tujuan : Menampilkan hasil operasi aritmatika (+, -, *, /, //, %) dari dua bilangan bulat

# Program operasi aritmatika dua bilangan bulat
# a = 25, b = 5

a = int(input("Masukkan bilangan bulat pertama: "))
b = int(input("Masukkan bilangan bulat kedua: "))

print(f"\nHasil operasi antara {a} dan {b}:")
print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} * {b} = {a * b}")

if b != 0:
    print(f"{a} / {b} = {a / b}")
    print(f"{a} // {b} = {a // b}")
    print(f"{a} % {b} = {a % b}")
else:
    print("Pembagian (/, //, %) tidak dapat dilakukan karena bilangan kedua adalah 0.")
```

```python
# Nama : Cindy Agatha Purba
# NIM : 4232501026
# Tanggal : 23 September 2026
# Tujuan : Menampilkan True atau False pada nilai kelulusan ujian

# Program pengecekan kelulusan berdasarkan nilai ujian
# Kategori lulus jika nilai >= 60

nilai = int(input("Masukkan nilai ujian (0-100): "))

hasil = nilai >= 60

print(f"\nNilai ujian: {nilai}")
print(f"Nilai ini termasuk kategori lulus (>= 60): {hasil}")
```

```python
# Nama : Cindy Agatha Purba
# NIM : 4232501026
# Tanggal : 23 September 2026
# Tujuan : Menentukan apakah seseorang boleh mengemudi sendiri

# Syarat: umur >= 17 DAN punya SIM

umur = int(input("Masukkan umur: "))
punya_sim_input = input("Punya SIM? (ya/tidak): ").strip().lower()
punya_sim = punya_sim_input == "ya"

# Gunakan operator logika (and)
boleh_mengemudi_sendiri = umur >= 17 and punya_sim

# Operator bitwise & TIDAK sama dengan 'and', dan precedence-nya lebih TINGGI
# daripada operator relasional (>=), sehingga tanpa tanda kurung hasilnya SALAH:
hasil_salah = umur >= 17 & punya_sim

# Dengan tanda kurung eksplisit, hasilnya menjadi BENAR:
hasil_benar = (umur >= 17) & punya_sim

print(f"\nUmur: {umur}")
print(f"Punya SIM: {punya_sim}")
print(f"Boleh mengemudi sendiri (pakai 'and'): {boleh_mengemudi_sendiri}")
print(f"Hasil TANPA kurung (umur >= 17 & punya_sim)  : {hasil_salah}   <-- SALAH, akibat precedence")
print(f"Hasil DENGAN kurung ((umur >= 17) & punya_sim): {hasil_benar}   <-- BENAR")
```

```python
# Nama : Cindy Agatha Purba
# NIM : 4232501026
# Tanggal : 23 September 2026
# Tujuan : Menghitung efisiensi termal unit pembangkit

# Rumus: efisiensi (%) = (energi output / energi input) * 100
# Satuan energi: MJ (Mega Joule)

energi_input = float(input("Masukkan energi input (MJ): "))
energi_output = float(input("Masukkan energi output (MJ): "))

if energi_input != 0:
    efisiensi = (energi_output / energi_input) * 100
    print(f"\nEnergi input   : {energi_input} MJ")
    print(f"Energi output  : {energi_output} MJ")
    print(f"Efisiensi termal: {efisiensi:.2f} %")
else:
    print("Energi input tidak boleh 0 (pembagian tidak terdefinisi).")
```

```python
# Nama : Cindy Agatha Purba
# NIM : 4232501026
# Tanggal : 23 September 2026
# Tujuan : Menentukan status "perlu maintenance" unit pembangkit

# Syarat: suhu boiler > 550°C ATAU beban operasi > 95%

suhu_boiler = float(input("Masukkan suhu boiler (°C): "))
beban_operasi = float(input("Masukkan beban operasi (%): "))

perlu_maintenance = (suhu_boiler > 550) or (beban_operasi > 95)

print(f"\nDengan suhu boiler {suhu_boiler}°C dan beban operasi {beban_operasi}%, "
      f"status unit ini perlu maintenance: {perlu_maintenance}")
```
