```python
#Belanja buku
jumlah_buku = 7
harga_buku = 18500
diskon_persen = 10
uang_budi = 150000
jumlah_orang = 4

#Total belanja, diskon, dan jumlah yang dibayar
total = jumlah_buku * harga_buku
diskon = total * diskon_persen // 100
bayar = total - diskon

#Uang kembalian
kembalian = uang_budi - bayar

#Patungan 4 orang (bilangan bulat)
per_orang = bayar // jumlah_orang
sisa = bayar % jumlah_orang

#Apakah uang cukup?
cukup = uang_budi >= bayar

print("Total        :", total)
print("Diskon       :", diskon)
print("Bayar        :", bayar)
print("Kembalian    :", kembalian)
print("Per orang    :", per_orang)
print("Sisa         :", sisa)
print("Uang cukup?  :", cukup)
```

```python
#PLTS atap rumah
jumlah_panel = 12
daya_panel = 450
jam_penyinaran = 5
efisiensi = 0.8
kebutuhan_harian = 25
tarif = 1444.70
hari_per_bulan = 30

#Energi per hari (kWh)
energi_harian = jumlah_panel * daya_panel * jam_penyinaran * efisiensi / 1000

#Energi per bulan dan nilai rupiahnya
energi_bulanan = energi_harian * hari_per_bulan
nilai_listrik = energi_bulanan * tarif

#Kekurangan energi per hari
kekurangan = kebutuhan_harian - energi_harian

#Apakah PLTS mencukupi?
mencukupi = energi_harian >= kebutuhan_harian

print("Energi harian (kWh) :", round(energi_harian, 1))
print("Energi bulanan (kWh):", round(energi_bulanan, 1))
print("Nilai listrik (Rp)  :", round(nilai_listrik, 1))
print("Kekurangan (kWh)    :", round(kekurangan, 1))
print("Mencukupi?          :", mencukupi)
```

```python
#Generator Diesel dan Rugi Kabel
daya_generator = 200
jam_operasi = 8
konsumsi = 0.25
harga_solar = 13500
kapasitas_tangki = 1000
arus = 20
hambatan = 0.5

#Energi yang dibangkitkan per hari (kWh)
energi = daya_generator * jam_operasi

#Kebutuhan solar per hari dan biayanya
solar = energi * konsumsi
biaya = solar * harga_solar

#Berapa hari tangki penuh cukup dan sisa solar
hari_cukup = kapasitas_tangki // solar
sisa_solar = kapasitas_tangki % solar

#Rugi daya pada kabel (P = I^2 x R)
rugi_kabel = arus ** 2 * hambatan

#Apakah rugi kabel tergolong tinggi?
rugi_tinggi = rugi_kabel > 150 and arus >= 20

print("Energi (kWh)     :", energi)
print("Solar (liter)    :", solar)
print("Biaya (Rp)       :", biaya)
print("Tangki cukup     :", hari_cukup, "hari")
print("Sisa solar (L)   :", sisa_solar)
print("Rugi kabel (W)   :", rugi_kabel)
print("Rugi tinggi?     :", rugi_tinggi)
```
