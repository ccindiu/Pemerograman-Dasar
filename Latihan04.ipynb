```python
1. Menentukan tahun kabisat
Sebuah tahun akan disebut tahun kabisat jika memenuhi salah satu dari dua kondisi tersebut:
a. Tahun tersebut habis dibagi 4, tetapi tidak habis dibagi 100
b. Tahun tersebut habis dibagi 400
Buatlah sebuah ekspresi logika kompleks untuk menentukan apakah suatu tahun merupakan tahun kabisat

Ekspresi logikanya dalam python adalah (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)
maka dapat dibuat
tahun = int(input("Masukkan tahun: "))

kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)
pesan = {True: "adalah tahun kabisat", False: "bukan tahun kabisat"}

print(tahun, pesan[kabisat])
```

```python
2. Buatlah program sederhana versi anda masing-masing dengan lingkup, fungsi input, fungsi output, dan operasi aritmatika
PI = 3.14  # variabel global (lingkup global)

def hitung_persegi_panjang():
    # variabel lokal (lingkup lokal, hanya ada di dalam fungsi)
    panjang = float(input("Masukkan panjang: "))   # fungsi input
    lebar = float(input("Masukkan lebar: "))

    luas = panjang * lebar                # operasi aritmatika
    keliling = 2 * (panjang + lebar)

    print("Luas     :", luas)             # fungsi output
    print("Keliling :", keliling)

hitung_persegi_panjang()
```
