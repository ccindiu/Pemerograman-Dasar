```python
#Modul KPI Pembangkit
energi = [5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
          6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
          5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062]

batubara = [2998, 3258, 3250, 3263, 3249, 3246, 2993, 2983, 3258, 3258,
            3253, 3249, 3244, 2996, 3009, 3257, 1950, 3234, 3261, 3262,
            3015, 2995, 3260, 3275, 3248, 3264, 3267, 2998, 3008, 3240]

jam_operasi = [24] * 30
jam_operasi[16] = 18   # hari ke-17 hanya 18 jam

def heat_rate(batubara_t, energi_mwh, nilai_kalor_kcal_kg=4400):
    """Heat rate (kcal/kWh)."""
    if energi_mwh <= 0:
        raise ValueError("energi_mwh harus positif")
    return batubara_t * nilai_kalor_kcal_kg / energi_mwh


def efisiensi_termal(heat_rate_kcal_kwh):
    """Efisiensi termal (%) dari heat rate."""
    return 860 / heat_rate_kcal_kwh * 100


def capacity_factor(energi_mwh, daya_terpasang_mw, jam=24):
    """Capacity factor (%) untuk suatu periode."""
    return energi_mwh / (daya_terpasang_mw * jam) * 100


def konsumsi_spesifik(batubara_t, energi_mwh):
    """Konsumsi batubara spesifik (t/MWh)."""
    return batubara_t / energi_mwh


def status_hari(efisiensi, jam_operasi):
    """TIDAK PENUH jika jam < 24, EFISIENSI RENDAH jika efisiensi < 36,3 %, selain itu NORMAL."""
    if jam_operasi < 24:
        return "TIDAK PENUH"
    if efisiensi < 36.3:
        return "EFISIENSI RENDAH"
    return "NORMAL"

def cetak_baris(hari, hr, efi, cf, ks, status):
    """Prosedur: satu baris tabel laporan."""
    print(f"{hari:>4} {hr:>7,.0f} {efi:>6.2f} {cf:>5.1f} {ks:>6.3f}  {status}")


def cetak_laporan(energi, batubara, jam_operasi, daya_terpasang_mw=330, nilai_kalor=4400):
    """Prosedur: cetak tabel harian. Mengembalikan daftar efisiensi dan daftar status."""
    print(f"{'Hari':>4} {'HR kcal':>7} {'Efi %':>6} {'CF %':>5} {'t/MWh':>6}  Status")
    efisiensi, daftar_status = [], []
    for i in range(len(energi)):
        hr = heat_rate(batubara[i], energi[i], nilai_kalor)
        efi = efisiensi_termal(hr)
        cf = capacity_factor(energi[i], daya_terpasang_mw, jam_operasi[i])
        ks = konsumsi_spesifik(batubara[i], energi[i])
        st = status_hari(efi, jam_operasi[i])
        efisiensi.append(efi)
        daftar_status.append(st)
        cetak_baris(i + 1, hr, efi, cf, ks, st)
    return efisiensi, daftar_status


def cetak_ringkasan(efisiensi, daftar_status, energi, daya_terpasang_mw=330):
    """Prosedur: rata-rata efisiensi, CF bulanan, jumlah hari per status, tiga hari terendah."""
    print(f"\nRata-rata efisiensi : {sum(efisiensi) / len(efisiensi):.2f} %")
    cf = capacity_factor(sum(energi), daya_terpasang_mw, jam=24 * len(energi))
    print(f"Capacity factor     : {cf:.1f} %")
    for s in ["NORMAL", "EFISIENSI RENDAH", "TIDAK PENUH"]:
        print(f"  {s:<17}: {daftar_status.count(s)} hari")
    print("Tiga hari paling tidak efisien:")
    for i in sorted(range(len(efisiensi)), key=lambda i: efisiensi[i])[:3]:
        print(f"  hari {i + 1}: {efisiensi[i]:.2f} %")

assert heat_rate(1, 2) == 2200
assert round(efisiensi_termal(2150), 2) == 40
assert round(capacity_factor(7920, 330)) == 100
assert konsumsi_spesifik(100, 200) == 0.5
assert status_hari(40, 18) == "TIDAK PENUH"
assert status_hari(36.0, 24) == "EFISIENSI RENDAH"
assert status_hari(36.3, 24) == "NORMAL"
print("Semua uji lulus")

efisiensi, daftar_status = cetak_laporan(energi, batubara, jam_operasi)
cetak_ringkasan(efisiensi, daftar_status, energi)

energi_u2 = [12150, 12080, 11020, 12210, 12095]
batubara_u2 = [5240, 5225, 4790, 5260, 5230]
jam_u2 = [24] * 5

efi_u2, status_u2 = cetak_laporan(energi_u2, batubara_u2, jam_u2,
                                  daya_terpasang_mw=600, nilai_kalor=5100)
cetak_ringkasan(efi_u2, status_u2, energi_u2, daya_terpasang_mw=600)
```

  Penjelasan:

  1. Program dimulai dengan tiga list: energi, batubara, dan jam_operasi. Ketiganya berpasangan menurut urutan, jadi posisi ke-0 di tiap list adalah hari pertama, posisi ke-1 hari kedua, dan seterusnya. Karena hari ke-17 hanya beroperasi 18 jam, jam_operasi[16] diubah menjadi 18 (indeks Python mulai dari 0).
  2. Lima fungsi menyimpan rumus, masing-masing satu tugas. heat_rate menghitung panas batubara per kWh listrik dan menolak energi nol atau negatif dengan ValueError. Hasilnya dioper ke efisiensi_termal yang mengubahnya menjadi persen lewat 860 ÷ heat rate × 100. capacity_factor membandingkan energi yang dihasilkan dengan kapasitas maksimal (daya × jam), dan konsumsi_spesifik menghitung ton batubara per MWh. status_hari memberi label dengan pemeriksaan berurutan: jam operasi dulu, baru efisiensi. Karena setiap cabang langsung return, pemeriksaan berhenti di kondisi pertama yang terpenuhi, sehingga hari 18 jam selalu "TIDAK PENUH" walau efisiensinya bagus.
  3. cetak_laporan mengulang dengan for i in range(len(energi)), jadi i berjalan dari 0 sampai hari terakhir. Di tiap putaran ia mengambil batubara[i], energi[i], dan jam_operasi[i], memanggil kelima fungsi, lalu mencetak satu baris tabel. Yang dicetak adalah i+1 agar hari tampil mulai dari 1. Status tiap hari juga disimpan ke list daftar_status, yang dikembalikan di akhir.
  4. cetak_ringkasan menghitung ulang efisiensi semua hari ke dalam satu list. Rata-rata didapat dari sum dibagi len. CF bulanan dihitung dari total energi dibagi kapasitas maksimal selama seluruh periode (daya × 24 × jumlah hari). Jumlah hari per status diambil dengan .count() pada daftar_status. Tiga hari paling tidak efisien dicari dengan mengurutkan nomor hari berdasarkan efisiensinya dari terkecil, lalu mengambil tiga yang pertama.
  5. Untuk Unit 2, program memanggil prosedur yang sama hanya dengan argumen berbeda (daya_mw=600, nilai_kalor=5100), karena rumusnya tidak ditulis ulang. Sebelum dipakai, fungsi-fungsi diuji dengan assert memakai angka kecil yang hasilnya bisa dihitung manual, misalnya heat_rate(1, 2) == 2200.
