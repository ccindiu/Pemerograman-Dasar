```python
#NAMA = CINDY AGATHA PURBA
#NIM = 4232501026
#TANGGAL = 9 NOVEMBER 2026
#MATA KULIAH = PEMROGRAMAN DASAR

#Soal 2. Energi Harian dan Capacity Factor PLTD
daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400

#a.energi harian (energi = daya_kw x jam)
energi = daya_kw * jam
print(f"Energi harian = {energi} kWh")

#b.Capacity factor
capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100
print(f"Capacity factor = {capacity_factor:.2f}%")
```

```python
#Soal 8. Pengosongan Tangki Bahan Bakar
volume = 100

while volume > 0:
    volume -= 20
    print(f"Volume: {volume} liter")
    if volume == 20:
        print("WARNING: Hampir kosong!")

print("TANGKI KOSONG - Pengosongan berhenti.")
```

Cara kerja program untuk soal nomor 8:
1. volume = 100: kondisi awal, tangki penuh.
2. while volume > 0: “selama tangki belum kosong, ulangi blok di bawahnya”. Sebelum setiap putaran, Python mengecek kondisi ini. Kalau salah, loop berhenti.
3. volume -= 20: singkatan dari volume = volume - 20.
4. print(...) dicetak setelah dikurangi, sesuai permintaan soal. Karena itu angka 100 tidak pernah muncul di output.
5. if volume == 20: berada di dalam loop, dicek setiap putaran. Hanya benar di putaran ke-4.
6. Baris terakhir tidak menjorok (di luar loop), sehingga hanya dijalankan sekali setelah loop selesai.

```python
#Soal 10. Pembacaan Suhu Bearing dengan Sensor Rusak
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

jumlah_valid = 0

for i, item in enumerate(data, start=1):
    try:
        nilai = float(item)
    except ValueError:
        print(f"Titik {i}: data tidak valid - pemeriksaan dihentikan")
        break

    if nilai > 70.0:
        print(f"Warning: Titik {i} suhu bearing tinggi!")

    jumlah_valid += 1

print(f"Jumlah data valid: {jumlah_valid}")
```

Cara kerja program untuk soal nomor 10:
1. jumlah_valid = 0: penghitung, mulai dari nol.
2. enumerate(data, start=1): mengambil data satu per satu sambil memberi nomor urut mulai dari 1. i adalah nomor titik, item adalah isi datanya (teks).
3. try: artinya “coba jalankan baris ini”.
 float("62.5") berhasil menjadi angka 62.5.
 float("N/A") gagal karena “N/A” bukan angka, sehingga Python memunculkan ValueError.
4. except ValueError: dijalankan hanya kalau ValueError muncul. Isinya mencetak pesan lalu break, yang langsung keluar dari loop. Data sesudahnya tidak diperiksa.
5. if nilai > 70.0: dicek hanya jika konversi berhasil. Tandanya > karena soal menyebut “melebihi”.
6. jumlah_valid += 1: ditambah setelah konversi berhasil, jadi hanya data valid yang terhitung.
7. Baris terakhir di luar loop, sehingga tercetak sekali setelah perulangan selesai (baik selesai normal maupun karena break).

```python
# PROGRAM PEMANTAUAN BEBAN LISTRIK HARIAN
# Masalah: Sebuah fasilitas mencatat beban listrik (kW) tiap jam.
# Program membaca data beban, mendeteksi anomali (beban terlalu tinggi / sensor rusak),
# lalu menghitung statistik dan estimasi konsumsi energi harian.

# Memuat
# 1. Data beban disimpan dalam list (bertipe teks, seperti hasil pembacaan sensor).
# 2. Perulangan for memeriksa data per jam.
# 3. try-except menangani data rusak (misal "N/A").
# 4. if-elif-else mengklasifikasikan status beban.
# 5. Setelah loop, hitung total energi (kWh), rata-rata, dan beban puncak.

# Batas beban (kW)
BATAS_NORMAL = 400
BATAS_WARNING = 500

# Data beban per jam (teks, karena dibaca dari sensor)
data_beban = ["320", "350", "410", "N/A", "520", "380", "-15", "450"]

beban_valid = []
jumlah_rusak = 0

print("=== PEMANTAUAN BEBAN LISTRIK HARIAN ===")

for jam, item in enumerate(data_beban, start=1):
    # Ubah teks menjadi angka; tangani data rusak
    try:
        beban = float(item)
        if beban < 0:
            raise ValueError("beban negatif")
    except ValueError:
        print(f"Jam {jam}: data tidak valid ({item}) - dilewati")
        jumlah_rusak += 1
        continue

    # Klasifikasi status beban
    if beban > BATAS_WARNING:
        status = "KRITIS"
    elif beban > BATAS_NORMAL:
        status = "WARNING"
    else:
        status = "NORMAL"

    beban_valid.append(beban)
    print(f"Jam {jam}: {beban:.1f} kW -> {status}")

# Ringkasan hasil (hanya jika ada data valid)
if len(beban_valid) > 0:
    energi_total = sum(beban_valid) * 1   # tiap data mewakili 1 jam
    rata_rata = energi_total / len(beban_valid)
    beban_puncak = max(beban_valid)

    print("\n=== RINGKASAN ===")
    print(f"Data valid       : {len(beban_valid)}")
    print(f"Data rusak       : {jumlah_rusak}")
    print(f"Energi total     : {energi_total:.2f} kWh")
    print(f"Beban rata-rata  : {rata_rata:.2f} kW")
    print(f"Beban puncak     : {beban_puncak:.2f} kW")
else:
    print("Tidak ada data valid.")
```

Cara kerja program untuk soal bonus:
1. Kita perlu menetapkan batas beban (400 dan 500 kW), siapkan data_beban (teks dari sensor), list kosong beban_valid, dan jumlah_rusak = 0.
2. Pada perulangan for: periksa data satu per satu dengan enumerate, nomor jam mulai dari 1.
3. Validasi try-except: ubah teks ke angka dengan float.
"N/A" gagal, ValueError muncul otomatis.
"-15" berhasil, tetapi negatif, jadi raise ValueError dipicu sendiri.
4. Tangani data rusak: cetak pesan, tambah jumlah_rusak, lalu continue untuk lanjut ke jam berikutnya.
5. Klasifikasi if-elif-else: di atas 500 = KRITIS, di atas 400 = WARNING, selain itu NORMAL.
6. Simpan dan cetak: masukkan beban ke beban_valid dengan append, lalu tampilkan hasilnya.
7. Ringkasan setelah loop: cek dulu ada data valid, lalu hitung total energi (sum), rata-rata (total dibagi len), dan puncak (max).
